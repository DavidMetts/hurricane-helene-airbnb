```python
import folium
import geopandas as gpd

# Load GeoJSON file
gdf = gpd.read_file("neighbourhoods.geojson")

# Create a Folium map centered around Asheville
m = folium.Map(location=[35.5951, -82.5515], zoom_start=12)

# Add interactive GeoJson layer with corrected field name
folium.GeoJson(
    gdf,
    name="Neighborhoods",
    tooltip=folium.GeoJsonTooltip(fields=["neighbourhood"], aliases=["Neighborhood:"])
).add_to(m)
```
